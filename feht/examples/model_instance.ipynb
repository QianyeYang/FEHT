```python
'''
This script provide a minimum example on how to integrate FEHT with a machine learning model.
'''
import feht, torch
```

```python
# using the default model 
model = feht.Model()
model = model.to('cuda')

# make some fake video data
# assume a scan video with 64 frames and 112x112 resolution (AFTER pre-processed)
video = torch.rand(1, 1, 64, 112, 112).to('cuda')
```

There are 3 types of output: 
* classification of heart presence
* center of the heart
* width and height of the heart, since the heart is basically a sphere, thus "radius" is used.

Note: The prediction of the center and the radius is a normalised value, you must know what is the original size of the video frames 

```python
pred_classes, pred_center, pred_radius = model(video)
pred_classes.shape, pred_center.shape, pred_radius.shape
```

However, the propose for implementing this tracking model is to be used in the pre-processing step. Most of the time the prediction of the bbox should be saved first and then used in other dataloaders. 

On the other hand, different videos varies in length. Thus a sliding window is needed during the inference, which could cause additional time and computation. 

Therefore we strongly recommend you to save the bbox prediction to disk and use them further, rather than using the model instance directly during your tasks.
