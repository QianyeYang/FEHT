```python
import os
import pandas as pd
import feht
```

```python
det_meta = feht.load_pkl('../../data/train-meta-example.pkl')
all_data = det_meta['test']
dataroot = "/home/engs2732/projects/DataSets/COCHE_site12_Bmode_videos"
```

```python
# make a table
# puid, suid, frame_path, height, width
index_data = {
    "puid": [],
    "suid": [],
    "frame_path": [],
    "height": [],
    "width": []
}
```

```python
for scan in all_data:
    puid = scan['puid']
    suid = scan['suid']
    for frame_path in scan['data_path']:
        index_data['puid'].append(puid)
        index_data['suid'].append(suid)
        index_data['frame_path'].append(
            os.path.join(dataroot, frame_path)
            )
        index_data['height'].append(scan['annotation_data'][0]['frame_shape'][0])
        index_data['width'].append(scan['annotation_data'][0]['frame_shape'][1])
```

```python
# # Create DataFrame
df = pd.DataFrame(index_data)
df.to_csv("../../data/inference_index_example.csv", index=False)
```
